```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def function(x):
    return x**2
```

```python
def der(x):
    return 2*x
```

```python
x = 2
learning_rate = 0.1
iterations = 50
```

```python
x_history = []
```

```python
for i in range(iterations):
    grad = der(x)
    x = x - learning_rate*function(x)
    x_history.append(x)
mini_val = function(x)
```

```python
print('minimum point : ',x)
print(mini_val)
```

```python
x_values = [i for i in range(-12,12)]
y_values = [function(i) for i in x_values]

plt.plot(x_values, y_values, label="f(x) = x²")
plt.scatter(x, mini_val, color="red",
            label="Minimum Point")
```
